# EDA: Default of Credit Card Clients

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/raw/UCI_Credit_Card.csv")
df.shape
```

```python
df.head()
```

## Пропуски и типы

```python
df.info()
```

```python
df.isnull().sum().sum()
```

## Баланс классов

```python
target = "default payment next month"
df[target].value_counts(normalize=True)
```

## Распределение возраста

```python
df["AGE"].hist(bins=30)
plt.xlabel("age")
plt.show()
```

## Лимит кредита по классам

```python
df.boxplot(column="LIMIT_BAL", by=target)
plt.show()
```

## Корреляция числовых признаков с таргетом

```python
numeric_df = df.select_dtypes("number")
numeric_df.corr()[target].sort_values(ascending=False)
```
